# DBSCAN

```python
# DBSCAN overcome the drawbacks of KMeans.
# Kmeans cannot handle the spherical data. DBSCAN can handle the spherical data.
# kmeans fails on spherical data because of centroid approach.
# DBscan works on spherical data because it works on the principle of nearest data points.
# kmeans can't handle the outliers. bcoz it will make the clusters with outliers also.
# DBScan can handle the outliers by making restriction on minimum number of data points needed to make the cluster.
```

```python
# Kmeans and DBscan both are unsupervised ML algorithm.
```

```python
# libraries
from sklearn.cluster import KMeans
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles # to generate spherical data
```

```python
# generate synthetic spherical data
x,y = make_circles(n_samples = 2000, factor = 0.4, noise = 0.05, random_state = 42)

plt.scatter(x[:,0],x[:,1])
```

```python
k = 2
kmeans_model = KMeans(n_clusters = k)
y_means = kmeans_model.fit_predict(x)

plt.scatter(x[:,0], x[:,1], c = y_means)
```

```python
dbscan_model = DBSCAN(eps = 0.2)
y_means = dbscan_model.fit_predict(x)

plt.scatter(x[:,0], x[:,1], c = y_means)
```
