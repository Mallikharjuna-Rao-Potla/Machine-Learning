# K-means Cluster

```python
# k-means

from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs # to design synthetic dataset
```

```python
# generate the synthetic dataset

x,y = make_blobs(n_samples = 300, centers = 3, n_features = 2, cluster_std = 2.0, random_state = 42)
#x[row, column]
#x[:,0] :
#x[:,1]:pick all the rows the data in 1st column
#

plt.scatter(x[:,0], x[:,1])
```

```python
# train our model
k = 3
kmean_model = KMeans(n_clusters = k, random_state = 42)
y_kmeans = kmean_model.fit_predict(x)
y_kmeans
```

```python
plt.scatter(x[:,0], x[:,1], c = y_kmeans)
```
